```python
import random
import math
import copy

maze = [
    ["S", ".", ".", "#"],
    ["#", "#", ".", "#"],
    [".", ".", ".", "."],
    ["#", ".", "#", "G"]
]

ROWS, COLS = 4, 4

# -------- FIND START AND GOAL --------
def find_pos(symbol):
    for i in range(ROWS):
        for j in range(COLS):
            if maze[i][j] == symbol:
                return (i, j)

start = find_pos("S")
goal = find_pos("G")

# -------- HEURISTIC (Manhattan Distance) --------
def heuristic(pos):
    return abs(pos[0] - goal[0]) + abs(pos[1] - goal[1])

# -------- VALID MOVE --------
def valid(x, y):
    return 0 <= x < ROWS and 0 <= y < COLS and maze[x][y] != "#"

# -------- GET NEIGHBORS --------
def neighbors(pos):
    x, y = pos
    moves = [(1,0),(-1,0),(0,1),(0,-1)]
    result = []

    for dx, dy in moves:
        nx, ny = x + dx, y + dy
        if valid(nx, ny):
            result.append((nx, ny))
    return result

# -------- RANDOM NEIGHBOR --------
def random_neighbor(pos):
    n = neighbors(pos)
    return random.choice(n) if n else pos

# -------- HILL CLIMBING --------
def hill_climbing():
    current = start
    path = [current]
    visited = set()

    while current != goal:
        visited.add(current)
        neigh = neighbors(current)

        best = current
        best_h = heuristic(current)

        for n in neigh:
            if n not in visited:
                h = heuristic(n)
                if h < best_h:
                    best = n
                    best_h = h

        if best == current:
            break  # stuck

        current = best
        path.append(current)

    return path, heuristic(current)

# -------- SIMULATED ANNEALING --------
def simulated_annealing():
    current = start
    path = [current]

    T = 10
    cooling = 0.95

    while current != goal and T > 0.1:
        next_state = random_neighbor(current)

        delta = heuristic(next_state) - heuristic(current)

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = next_state
            path.append(current)

        T *= cooling

    return path, heuristic(current)

# -------- MAIN --------
hc_path, hc_cost = hill_climbing()
sa_path, sa_cost = simulated_annealing()

print("Hill Climbing Path:", hc_path)
print("Final Heuristic:", hc_cost)

print("\nSimulated Annealing Path:", sa_path)
print("Final Heuristic:", sa_cost)
```
