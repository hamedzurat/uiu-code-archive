```python
import random
import math

# -------- JOB DATA --------
processing_time = [2, 1, 3, 2]
N = len(processing_time)

# -------- RANDOM SOLUTION --------
def random_solution():
    sol = list(range(N))
    random.shuffle(sol)
    return sol

# -------- COST FUNCTION (makespan / total completion time) --------
def cost(sol):
    time = 0
    total = 0

    for job in sol:
        time += processing_time[job]
        total += time

    return total

# -------- NEIGHBORS (swap jobs) --------
def neighbors(sol):
    result = []
    for i in range(N):
        for j in range(i + 1, N):
            new_sol = sol[:]
            new_sol[i], new_sol[j] = new_sol[j], new_sol[i]
            result.append(new_sol)
    return result

# -------- RANDOM NEIGHBOR --------
def random_neighbor(sol):
    i, j = random.sample(range(N), 2)
    new_sol = sol[:]
    new_sol[i], new_sol[j] = new_sol[j], new_sol[i]
    return new_sol

# -------- HILL CLIMBING --------
def hill_climbing():
    current = random_solution()
    current_cost = cost(current)

    while True:
        neigh = neighbors(current)

        best = current
        best_cost = current_cost

        for n in neigh:
            c = cost(n)
            if c < best_cost:
                best = n
                best_cost = c

        if best_cost >= current_cost:
            break

        current = best
        current_cost = best_cost

    return current, current_cost

# -------- SIMULATED ANNEALING --------
def simulated_annealing():
    current = random_solution()
    current_cost = cost(current)

    best = current[:]
    best_cost = current_cost

    T = 10
    cooling = 0.95

    while T > 0.1:
        next_state = random_neighbor(current)
        next_cost = cost(next_state)

        delta = next_cost - current_cost

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = next_state
            current_cost = next_cost

            if current_cost < best_cost:
                best = current[:]
                best_cost = current_cost

        T *= cooling

    return best, best_cost

# -------- MAIN --------
hc_sol, hc_cost = hill_climbing()
sa_sol, sa_cost = simulated_annealing()

print("Hill Climbing:")
print("Job Order:", hc_sol)
print("Total Completion Time:", hc_cost)

print("\nSimulated Annealing:")
print("Job Order:", sa_sol)
print("Total Completion Time:", sa_cost)
```
