```python
import random
import math

# -------- DISTANCE MATRIX --------
dist = [
    [0, 10, 15, 20, 25],
    [10, 0, 35, 25, 30],
    [15, 35, 0, 30, 20],
    [20, 25, 30, 0, 15],
    [25, 30, 20, 15, 0]
]

N = 4  # customers (1-4)

# -------- RANDOM SOLUTION --------
def random_solution():
    sol = list(range(1, N + 1))
    random.shuffle(sol)
    return sol

# -------- COST FUNCTION --------
def cost(sol):
    total = 0
    prev = 0  # depot

    for city in sol:
        total += dist[prev][city]
        prev = city

    total += dist[prev][0]  # return to depot
    return total

# -------- NEIGHBORS --------
def neighbors(sol):
    res = []
    for i in range(len(sol)):
        for j in range(i + 1, len(sol)):
            new_sol = sol[:]
            new_sol[i], new_sol[j] = new_sol[j], new_sol[i]
            res.append(new_sol)
    return res

# -------- RANDOM NEIGHBOR --------
def random_neighbor(sol):
    i, j = random.sample(range(len(sol)), 2)
    new_sol = sol[:]
    new_sol[i], new_sol[j] = new_sol[j], new_sol[i]
    return new_sol

# -------- HILL CLIMBING --------
def hill_climbing():
    current = random_solution()
    current_cost = cost(current)

    while True:
        neigh = neighbors(current)
        best = current
        best_cost = current_cost

        for n in neigh:
            c = cost(n)
            if c < best_cost:
                best = n
                best_cost = c

        if best_cost >= current_cost:
            break

        current = best
        current_cost = best_cost

    return current, current_cost

# -------- SIMULATED ANNEALING --------
def simulated_annealing():
    current = random_solution()
    current_cost = cost(current)

    best = current[:]
    best_cost = current_cost

    T = 10
    cooling = 0.95

    while T > 0.1:
        next_state = random_neighbor(current)
        next_cost = cost(next_state)

        delta = next_cost - current_cost

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = next_state
            current_cost = next_cost

            if current_cost < best_cost:
                best = current[:]
                best_cost = current_cost

        T *= cooling

    return best, best_cost

# -------- MAIN --------
hc, hc_cost = hill_climbing()
sa, sa_cost = simulated_annealing()

print("Hill Climbing:", hc, hc_cost)
print("Simulated Annealing:", sa, sa_cost)
```
