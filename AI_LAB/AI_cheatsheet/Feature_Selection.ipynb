```python
import random
import math

features = [0.8, 0.4, 0.9, 0.3, 0.7]
N = len(features)

# -------- RANDOM STATE --------
def random_state():
    while True:
        sol = [random.randint(0, 1) for _ in range(N)]
        if sum(sol) > 0:
            return sol

# -------- FITNESS --------
def fitness(sol):
    return sum(features[i] for i in range(N) if sol[i] == 1)

# -------- NEIGHBORS --------
def neighbors(sol):
    res = []
    for i in range(N):
        new_sol = sol[:]
        new_sol[i] = 1 - new_sol[i]
        if sum(new_sol) > 0:
            res.append(new_sol)
    return res

# -------- RANDOM NEIGHBOR --------
def random_neighbor(sol):
    i = random.randint(0, N - 1)
    new_sol = sol[:]
    new_sol[i] = 1 - new_sol[i]
    if sum(new_sol) == 0:
        return sol
    return new_sol

# -------- HILL CLIMBING --------
def hill_climbing():
    current = random_state()
    current_fit = fitness(current)

    while True:
        neigh = neighbors(current)
        best = current
        best_fit = current_fit

        for n in neigh:
            f = fitness(n)
            if f > best_fit:
                best = n
                best_fit = f

        if best_fit <= current_fit:
            break

        current = best
        current_fit = best_fit

    return current, current_fit

# -------- SIMULATED ANNEALING --------
def simulated_annealing():
    current = random_state()
    current_fit = fitness(current)

    best = current[:]
    best_fit = current_fit

    T = 10
    cooling = 0.95

    while T > 0.1:
        next_state = random_neighbor(current)
        next_fit = fitness(next_state)

        delta = next_fit - current_fit

        if delta > 0 or random.random() < math.exp(delta / T):
            current = next_state
            current_fit = next_fit

            if current_fit > best_fit:
                best = current[:]
                best_fit = current_fit

        T *= cooling

    return best, best_fit

# -------- MAIN --------
hc, hc_fit = hill_climbing()
sa, sa_fit = simulated_annealing()

print("Hill Climbing:", hc, hc_fit)
print("Simulated Annealing:", sa, sa_fit)
```
