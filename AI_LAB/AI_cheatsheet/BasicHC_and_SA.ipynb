```python
import random

def h(pos,goal):
  return abs(pos[0]-goal[0])+abs(pos[1]-goal[1])

def stochastic_hill_climbing(maze, start, goal, max_iterations):
  current_pos = start
  current_cost=h(current_pos,goal)

  path = [current_pos]

  moves=[(0,1),(1,0),(-1,0),(0,-1)]

  for i in range (max_iterations):
    choice = random.choice(moves)
    new_pos = (current_pos[0]+choice[0],current_pos[1]+choice[1])

    if new_pos==goal : return path,True

    if 0<=new_pos[0]<len(maze) and 0<=new_pos[1]<len(maze[0]) and maze[new_pos[0]][new_pos[1]]==0:
      new_cost = h(new_pos,goal)
      if new_cost < current_cost:
        current_pos = new_pos
        current_cost = new_cost
        path.append(current_pos)
  return path,False
```

```python
import random
import math

def h(pos,goal):
  return abs(pos[0]-goal[0])+abs(pos[1]-goal[1])

def simulated_annealing(maze, start, goal, max_iterations):
  current_pos = start
  current_cost=h(current_pos,goal)

  temperature = 10
  cooling_rate = 0.95

  path = [current_pos]

  moves=[(0,1),(1,0),(-1,0),(0,-1)]

  for i in range (max_iterations):
    choice = random.choice(moves)
    new_pos = (current_pos[0]+choice[0],current_pos[1]+choice[1])

    if new_pos==goal : return path,True

    if 0<=new_pos[0]<len(maze) and 0<=new_pos[1]<len(maze[0]) and maze[new_pos[0]][new_pos[1]]==0:
      new_cost = h(new_pos,goal)
      diff = new_cost-current_cost
      if diff<0:
        current_pos = new_pos
        current_cost = new_cost
        path.append(current_pos)
      else:
        probability = math.exp(-diff/temperature)
        if random.uniform(0,1)<probability:
          current_pos = new_pos
          current_cost = new_cost
          path.append(current_pos)

      temperature *= cooling_rate


  return path,False
```

```python


maze = [
    [0, 0, 0, 0, 0],
    [1, 1, 0, 1, 0],
    [0, 0, 0, 1, 0],
    [1, 1, 0, 1, 0],
    [0, 0, 0, 0, 0],
]
start = (0, 0)
goal = (4, 4)
max_iterations = 10000

```

```python
path, success = stochastic_hill_climbing(maze, start, goal, max_iterations)

if success:
    print("Path found:", path)
else:
    print("No path found within the specified number of iterations.")
```

```python
path, success = simulated_annealing(maze, start, goal, max_iterations)

if success:
    print("Path found:", path)
else:
    print("No path found within the specified number of iterations.")
```
