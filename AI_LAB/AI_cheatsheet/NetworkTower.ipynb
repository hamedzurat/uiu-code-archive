```python
import random
import math

# -------------------------
# Problem Setup
# -------------------------

grid = [
    [5, 1, 2, 3],
    [9, "#", 1, 4],
    [2, 8, 6, 1],
    [1, 3, 7, 2]
]

K = 4

rows = len(grid)
cols = len(grid[0])

# -------------------------
# Helper functions
# -------------------------

def is_valid(cell):
    x, y = cell
    return 0 <= x < rows and 0 <= y < cols and grid[x][y] != "#"


def get_value(cell):
    x, y = cell
    return grid[x][y]


def adjacency_penalty(state):
    penalty = 0
    state = list(state)

    for i in range(len(state)):
        for j in range(i + 1, len(state)):
            x1, y1 = state[i]
            x2, y2 = state[j]

            if abs(x1 - x2) + abs(y1 - y2) == 1:
                penalty += 3

    return penalty


def score(state):
    total = sum(get_value(c) for c in state)
    return total - adjacency_penalty(state)


# -------------------------
# Generate initial state
# -------------------------

def random_state():
    cells = [(i, j) for i in range(rows) for j in range(cols) if grid[i][j] != "#"]
    return tuple(random.sample(cells, K))


# -------------------------
# Neighbor function
# -------------------------

def get_neighbor(state):
    state = list(state)

    all_cells = [(i, j) for i in range(rows) for j in range(cols) if grid[i][j] != "#"]

    remove_idx = random.randint(0, K - 1)
    removed = state[remove_idx]

    candidates = [c for c in all_cells if c not in state]

    if not candidates:
        return state

    new_cell = random.choice(candidates)

    state[remove_idx] = new_cell
    return tuple(state)


# -------------------------
# Hill Climbing
# -------------------------

def hill_climb(iterations=1000):
    current = random_state()
    current_score = score(current)

    for _ in range(iterations):
        neighbor = get_neighbor(current)
        neighbor_score = score(neighbor)

        if neighbor_score > current_score:
            current = neighbor
            current_score = neighbor_score

    return current, current_score


# -------------------------
# Simulated Annealing
# -------------------------

def simulated_annealing(T=1000, cooling=0.99):
    current = random_state()
    current_score = score(current)

    best = current
    best_score = current_score

    temperature = T

    while temperature > 1:
        neighbor = get_neighbor(current)
        neighbor_score = score(neighbor)

        delta = neighbor_score - current_score

        if delta > 0 or random.random() < math.exp(delta / temperature):
            current = neighbor
            current_score = neighbor_score

            if current_score > best_score:
                best = current
                best_score = current_score

        temperature *= cooling

    return best, best_score


# -------------------------
# Run both
# -------------------------

best_hc, score_hc = hill_climb()
best_sa, score_sa = simulated_annealing()

print("Hill Climbing Result:")
print("Score:", score_hc)
print("State:", best_hc)

print("\nSimulated Annealing Result:")
print("Score:", score_sa)
print("State:", best_sa)
```
