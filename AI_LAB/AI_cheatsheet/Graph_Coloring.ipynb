```python
import random
import math

# -------- GRAPH --------
edges = [(0,1),(0,2),(1,2),(1,3),(2,3)]
N = 4
COLORS = [0,1,2]

# -------- RANDOM STATE --------
def random_state():
    return [random.choice(COLORS) for _ in range(N)]

# -------- CONFLICT COUNT --------
def conflicts(state):
    c = 0
    for u, v in edges:
        if state[u] == state[v]:
            c += 1
    return c

# -------- NEIGHBOR (change one node color) --------
def neighbors(state):
    result = []
    for i in range(N):
        for color in COLORS:
            if state[i] != color:
                new_state = state[:]
                new_state[i] = color
                result.append(new_state)
    return result

# -------- RANDOM NEIGHBOR --------
def random_neighbor(state):
    i = random.randint(0, N - 1)
    new_state = state[:]
    new_state[i] = random.choice(COLORS)
    return new_state

# -------- HILL CLIMBING --------
def hill_climbing():
    current = random_state()
    current_conf = conflicts(current)

    while True:
        neigh = neighbors(current)

        best = current
        best_conf = current_conf

        for n in neigh:
            c = conflicts(n)
            if c < best_conf:
                best = n
                best_conf = c

        if best_conf >= current_conf:
            break

        current = best
        current_conf = best_conf

    return current, current_conf

# -------- SIMULATED ANNEALING --------
def simulated_annealing():
    current = random_state()
    current_conf = conflicts(current)

    best = current[:]
    best_conf = current_conf

    T = 10
    cooling = 0.95

    while T > 0.1:
        next_state = random_neighbor(current)
        next_conf = conflicts(next_state)

        delta = next_conf - current_conf

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = next_state
            current_conf = next_conf

            if current_conf < best_conf:
                best = current[:]
                best_conf = current_conf

        T *= cooling

    return best, best_conf

# -------- MAIN --------
hc_state, hc_conf = hill_climbing()
sa_state, sa_conf = simulated_annealing()

print("Hill Climbing:")
print("Coloring:", hc_state)
print("Conflicts:", hc_conf)

print("\nSimulated Annealing:")
print("Coloring:", sa_state)
print("Conflicts:", sa_conf)
```
