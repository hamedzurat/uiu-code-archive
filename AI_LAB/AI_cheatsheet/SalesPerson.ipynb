TRAVELLING SALES PERSON

```python
import random
import math

# -------- COMMON FUNCTIONS --------
def random_solution(tsp):
    sol = list(range(len(tsp)))
    random.shuffle(sol)
    return sol

def cost(tsp, sol):
    c = 0
    for i in range(len(sol) - 1):
        c += tsp[sol[i]][sol[i + 1]]
    c += tsp[sol[-1]][sol[0]]  # return to start
    return c

def get_neighbors(sol):
    neighbors = []
    n = len(sol)
    for i in range(n):
        for j in range(i + 1, n):
            new_sol = sol[:]
            new_sol[i], new_sol[j] = new_sol[j], new_sol[i]
            neighbors.append(new_sol)
    return neighbors

def random_neighbor(sol):
    i, j = random.sample(range(len(sol)), 2)
    new_sol = sol[:]
    new_sol[i], new_sol[j] = new_sol[j], new_sol[i]
    return new_sol

# -------- HILL CLIMBING --------
def hill_climbing(tsp):
    current = random_solution(tsp)
    current_cost = cost(tsp, current)

    while True:
        neighbors = get_neighbors(current)
        best = current
        best_cost = current_cost

        for n in neighbors:
            c = cost(tsp, n)
            if c < best_cost:
                best = n
                best_cost = c

        if best_cost >= current_cost:
            break

        current = best
        current_cost = best_cost

    return current, current_cost

# -------- SIMULATED ANNEALING --------
def simulated_annealing(tsp):
    current = random_solution(tsp)
    current_cost = cost(tsp, current)

    T = 1000
    cooling = 0.995

    best = current[:]
    best_cost = current_cost

    while T > 1:
        next_sol = random_neighbor(current)
        next_cost = cost(tsp, next_sol)

        delta = next_cost - current_cost

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = next_sol
            current_cost = next_cost

            if current_cost < best_cost:
                best = current[:]
                best_cost = current_cost

        T *= cooling

    return best, best_cost

# -------- MAIN --------
if __name__ == "__main__":
    tsp = [
        [0, 400, 500, 300],
        [400, 0, 300, 500],
        [500, 300, 0, 400],
        [300, 500, 400, 0]
    ]

    hc_path, hc_cost = hill_climbing(tsp)
    sa_path, sa_cost = simulated_annealing(tsp)

    print("Hill Climbing:")
    print("Path:", hc_path)
    print("Cost:", hc_cost)

    print("\nSimulated Annealing:")
    print("Path:", sa_path)
    print("Cost:", sa_cost)
```
