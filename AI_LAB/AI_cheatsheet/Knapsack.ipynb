```python
import random
import math

# -------- DATA --------
values = [60, 100, 120]
weights = [10, 20, 30]
CAPACITY = 50
N = len(values)

# -------- RANDOM STATE --------
def random_state():
    return [random.randint(0, 1) for _ in range(N)]

# -------- FITNESS FUNCTION --------
def fitness(state):
    total_value = 0
    total_weight = 0

    for i in range(N):
        if state[i] == 1:
            total_value += values[i]
            total_weight += weights[i]

    # penalize overweight solutions
    if total_weight > CAPACITY:
        return 0

    return total_value

# -------- NEIGHBOR GENERATION --------
def neighbors(state):
    result = []
    for i in range(N):
        new_state = state[:]
        new_state[i] = 1 - new_state[i]  # flip bit
        result.append(new_state)
    return result

# -------- RANDOM NEIGHBOR --------
def random_neighbor(state):
    i = random.randint(0, N - 1)
    new_state = state[:]
    new_state[i] = 1 - new_state[i]
    return new_state

# -------- HILL CLIMBING --------
def hill_climbing():
    current = random_state()
    current_fit = fitness(current)

    while True:
        neigh = neighbors(current)

        best = current
        best_fit = current_fit

        for n in neigh:
            f = fitness(n)
            if f > best_fit:
                best = n
                best_fit = f

        if best_fit <= current_fit:
            break

        current = best
        current_fit = best_fit

    return current, current_fit

# -------- SIMULATED ANNEALING --------
def simulated_annealing():
    current = random_state()
    current_fit = fitness(current)

    best = current[:]
    best_fit = current_fit

    T = 10
    cooling = 0.95

    while T > 0.1:
        next_state = random_neighbor(current)
        next_fit = fitness(next_state)

        delta = next_fit - current_fit

        if delta > 0 or random.random() < math.exp(delta / T):
            current = next_state
            current_fit = next_fit

            if current_fit > best_fit:
                best = current[:]
                best_fit = current_fit

        T *= cooling

    return best, best_fit

# -------- MAIN --------
hc_sol, hc_val = hill_climbing()
sa_sol, sa_val = simulated_annealing()

def decode(sol):
    items = [i for i in range(N) if sol[i] == 1]
    weight = sum(weights[i] for i in items)
    value = sum(values[i] for i in items)
    return items, weight, value

print("Hill Climbing:")
items, w, v = decode(hc_sol)
print("Items:", items)
print("Weight:", w, "Value:", v)

print("\nSimulated Annealing:")
items, w, v = decode(sa_sol)
print("Items:", items)
print("Weight:", w, "Value:", v)
```
