```python
import random
import math
import copy

GOAL = [
    [1,2,3],
    [4,5,6],
    [7,8,0]
]

# -------- HEURISTIC (Manhattan Distance) --------
def heuristic(state):
    dist = 0
    for i in range(3):
        for j in range(3):
            val = state[i][j]
            if val != 0:
                goal_x = (val - 1) // 3
                goal_y = (val - 1) % 3
                dist += abs(i - goal_x) + abs(j - goal_y)
    return dist

# -------- FIND BLANK --------
def find_zero(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j

# -------- GENERATE NEIGHBORS --------
def get_neighbors(state):
    neighbors = []
    x, y = find_zero(state)
    moves = [(-1,0),(1,0),(0,-1),(0,1)]

    for dx, dy in moves:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = copy.deepcopy(state)
            new_state[x][y], new_state[nx][ny] = new_state[nx][ny], new_state[x][y]
            neighbors.append(new_state)
    return neighbors

def random_neighbor(state):
    return random.choice(get_neighbors(state))

# -------- HILL CLIMBING --------
def hill_climbing(start):
    current = start
    current_h = heuristic(current)

    while True:
        neighbors = get_neighbors(current)
        best = current
        best_h = current_h

        for n in neighbors:
            h = heuristic(n)
            if h < best_h:
                best = n
                best_h = h

        if best_h >= current_h:
            break

        current = best
        current_h = best_h

    return current, current_h

# -------- SIMULATED ANNEALING --------
def simulated_annealing(start):
    current = start
    current_h = heuristic(current)

    T = 10
    cooling = 0.99

    best = current
    best_h = current_h

    while T > 0.001:
        next_state = random_neighbor(current)
        next_h = heuristic(next_state)

        delta = next_h - current_h

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = next_state
            current_h = next_h

            if current_h < best_h:
                best = current
                best_h = current_h

        T *= cooling

    return best, best_h

# -------- MAIN --------
if __name__ == "__main__":
    start = [
        [1,2,3],
        [4,0,6],
        [7,5,8]
    ]

    hc_state, hc_h = hill_climbing(start)
    sa_state, sa_h = simulated_annealing(start)

    print("Hill Climbing Result:")
    print(hc_state)
    print("Heuristic:", hc_h)

    print("\nSimulated Annealing Result:")
    print(sa_state)
    print("Heuristic:", sa_h)
```
