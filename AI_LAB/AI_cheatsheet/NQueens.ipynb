```python
import random
import math

N = 4

# -------- COMMON FUNCTIONS --------

# random initial state
def random_state():
    return [random.randint(0, N - 1) for _ in range(N)]

# count conflicts
def conflicts(state):
    count = 0
    for i in range(N):
        for j in range(i + 1, N):
            if state[i] == state[j]:  # same row
                count += 1
            if abs(state[i] - state[j]) == abs(i - j):  # diagonal
                count += 1

    return count

# generate neighbors (move one queen)
def get_neighbors(state):
    neighbors = []
    for col in range(N):
        for row in range(N):
            if state[col] != row:
                new_state = state[:]
                new_state[col] = row
                neighbors.append(new_state)
    return neighbors

# random neighbor (for SA)
def random_neighbor(state):
    col = random.randint(0, N - 1)
    row = random.randint(0, N - 1)
    new_state = state[:]
    new_state[col] = row
    return new_state

# -------- HILL CLIMBING --------
def hill_climbing():
    current = random_state()
    current_conf = conflicts(current)

    while True:
        neighbors = get_neighbors(current)
        best = current
        best_conf = current_conf

        for n in neighbors:
            c = conflicts(n)
            if c < best_conf:
                best = n
                best_conf = c

        if best_conf >= current_conf:
            break

        current = best
        current_conf = best_conf

    return current, current_conf

# -------- SIMULATED ANNEALING --------
def simulated_annealing():
    current = random_state()
    current_conf = conflicts(current)

    T = 100
    cooling = 0.95

    best = current[:]
    best_conf = current_conf

    while T > 0.1:
        next_state = random_neighbor(current)
        next_conf = conflicts(next_state)

        delta = next_conf - current_conf

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = next_state
            current_conf = next_conf

            if current_conf < best_conf:
                best = current[:]
                best_conf = current_conf

        T *= cooling

    return best, best_conf

# -------- MAIN --------
if __name__ == "__main__":
    hc_state, hc_conf = hill_climbing()
    sa_state, sa_conf = simulated_annealing()

    print("Hill Climbing:")
    print("State:", hc_state)
    print("Conflicts:", hc_conf)

    print("\nSimulated Annealing:")
    print("State:", sa_state)
    print("Conflicts:", sa_conf)
```
