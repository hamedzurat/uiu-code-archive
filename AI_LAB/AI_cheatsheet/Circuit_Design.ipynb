```python
import random
import math

# -------- COST MATRIX --------
cost = [
    [0, 2, 9, 10],
    [1, 0, 6, 4],
    [15, 7, 0, 8],
    [6, 3, 12, 0]
]

N = 4

# -------- RANDOM SOLUTION --------
def random_solution():
    sol = list(range(N))
    random.shuffle(sol)
    return sol

# -------- TOTAL COST --------
def total_cost(sol):
    c = 0
    for i in range(N - 1):
        c += cost[sol[i]][sol[i + 1]]
    return c

# -------- NEIGHBORS (swap two components) --------
def neighbors(sol):
    result = []
    for i in range(N):
        for j in range(i + 1, N):
            new_sol = sol[:]
            new_sol[i], new_sol[j] = new_sol[j], new_sol[i]
            result.append(new_sol)
    return result

# -------- RANDOM NEIGHBOR --------
def random_neighbor(sol):
    i, j = random.sample(range(N), 2)
    new_sol = sol[:]
    new_sol[i], new_sol[j] = new_sol[j], new_sol[i]
    return new_sol

# -------- HILL CLIMBING --------
def hill_climbing():
    current = random_solution()
    current_cost = total_cost(current)

    while True:
        neigh = neighbors(current)

        best = current
        best_cost = current_cost

        for n in neigh:
            c = total_cost(n)
            if c < best_cost:
                best = n
                best_cost = c

        if best_cost >= current_cost:
            break

        current = best
        current_cost = best_cost

    return current, current_cost

# -------- SIMULATED ANNEALING --------
def simulated_annealing():
    current = random_solution()
    current_cost = total_cost(current)

    best = current[:]
    best_cost = current_cost

    T = 10
    cooling = 0.95

    while T > 0.1:
        next_state = random_neighbor(current)
        next_cost = total_cost(next_state)

        delta = next_cost - current_cost

        if delta < 0 or random.random() < math.exp(-delta / T):
            current = next_state
            current_cost = next_cost

            if current_cost < best_cost:
                best = current[:]
                best_cost = current_cost

        T *= cooling

    return best, best_cost

# -------- MAIN --------
hc_sol, hc_cost = hill_climbing()
sa_sol, sa_cost = simulated_annealing()

print("Hill Climbing:")
print("Order:", hc_sol)
print("Delay:", hc_cost)

print("\nSimulated Annealing:")
print("Order:", sa_sol)
print("Delay:", sa_cost)
```
