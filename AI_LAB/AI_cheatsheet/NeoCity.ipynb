```python
import heapq as hq
from math import inf


def h(a, b):
    return abs(a[0] - b[0]) + abs(a[1] - b[1])


def a_star(grid, moves):

    rows = len(grid)
    cols = len(grid[0])

    start, goal = None, None

    # find start and goal
    for i in range(rows):
        for j in range(cols):
            if grid[i][j] == "S":
                start = (i, j)
            if grid[i][j] == "G":
                goal = (i, j)

    # FIX 1: correct 2D parent + distance
    parent = [[(-1, -1) for _ in range(cols)] for _ in range(rows)]
    distance = [[inf for _ in range(cols)] for _ in range(rows)]

    distance[start[0]][start[1]] = 0

    q = []
    hq.heappush(q, (0 + h(start, goal), start))

    while q:
        cost, node = hq.heappop(q)
        x, y = node

        if node == goal:
            # FIX 2: reconstruct path before returning
            path = []
            cx, cy = goal
            while (cx, cy) != (-1, -1) and (cx, cy) != start:
                path.append((cx, cy))
                cx, cy = parent[cx][cy]
            path.append(start)
            path.reverse()

            return len(path), path, distance[goal[0]][goal[1]], True

        for dx, dy in moves:
            nx, ny = x + dx, y + dy

            if 0 <= nx < rows and 0 <= ny < cols and grid[nx][ny] != "#":

                new_cost = distance[x][y] + 1

                if grid[nx][ny] == "W":
                    new_cost += 2

                if new_cost < distance[nx][ny]:
                    distance[nx][ny] = new_cost
                    parent[nx][ny] = (x, y)
                    hq.heappush(q, (new_cost + h((nx, ny), goal), (nx, ny)))

    return -1, [], inf, False
```

```python
grid = [
    "S..W.",
    ".##W.",
    ".W...",
    "..##.",
    "...G."
]


rows = len(grid)
cols = len(grid[0])

moves= [
    (0,1),
    (1,0),
    (-1,0),
    (0,-1)
]

count, path, cost, success = a_star(grid,moves)

print(f"Count: {count}")
print(f"Path: {path}")
print(f"Cost: {cost}")
print(f"Success: {success}")
```
