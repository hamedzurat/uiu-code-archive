```python
import random

def heuristic(node, goal):
    return abs(node[0] - goal[0]) + abs(node[1] - goal[1])

def stochastic_hill_climbing(maze, start, goal, max_iterations):
    current_pos = start
    current_cost = heuristic(current_pos, goal)
    path = [current_pos]


    moves = [(0, 1), (1, 0), (0, -1), (-1, 0)]

    for i in range(max_iterations):
        if current_pos == goal:
            return path, True
        move = random.choice(moves)
        new_pos = (current_pos[0] + move[0], current_pos[1] + move[1])

        # checck if validity of neightbor state
        if(0<= new_pos[0] < len(maze)) and (0 <= new_pos[1] < len(maze[0])) and (maze[new_pos[0]][new_pos[1]] == 0):
            new_cost = heuristic(new_pos, goal)
            diff = new_cost - current_cost

            if diff < 0:
                current_pos = new_pos
                current_cost = new_cost
                path.append(current_pos)

    return path, False
    
```

```python
maze = [
    [0, 0, 0, 0, 0],
    [1, 1, 0, 1, 0],
    [0, 0, 0, 1, 0],
    [1, 1, 0, 1, 0],
    [0, 0, 0, 0, 0],
]
start = (0, 0)
goal = (4, 4)
max_iterations = 100
```

```python
path, success = stochastic_hill_climbing(maze, start, goal, max_iterations)

if success:
    print("Path found:", path)
else:
    print("No path found within the specified number of iterations.")
```
