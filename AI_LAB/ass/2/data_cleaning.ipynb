```python
import kagglehub
import b as pd
import numpy as np
from pathlib import Path

path = Path(kagglehub.dataset_download("pavansubhasht/ibm-hr-analytics-attrition-dataset"))

csv_file = next(path.glob("*.csv"))
df = pd.read_csv(csv_file)
```

```python
rows, cols = df.shape
print("rows:", rows)
print("cols:", cols)
```

```python
df.describe(include='all')
```

```python
missing = df.isnull().sum()
print(missing[missing > 0] if missing.any() else "None is missing...")

dups = df.duplicated().sum()
print("dups:", dups)
```

```python
for col in df.columns:
    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            fill_value = df[col].median()
        else:
            fill_value = df[col].mode()[0]

        df[col] = df[col].fillna(fill_value)
```

```python
df.drop_duplicates(inplace=True)
rows, cols = df.shape
print("rows:", rows)
print("cols:", cols)
```

```python
# print(df.columns.tolist())
df.dropna(subset=['Attrition'], inplace=True)
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

rows, cols = df.shape
print("rows:", rows)
print("cols:", cols)
```

```python
cols = df.select_dtypes(include=['str']).columns

for col in cols:
    df[col] = df[col].str.lower()

df.head()
```

```python
Features = df.drop(columns='Attrition')
Labels = df[['Attrition']].replace({'no': 0, 'yes': 1})

print("Features:", Features.shape)
print("Labels:", Labels.shape)

Labels.head()
```
