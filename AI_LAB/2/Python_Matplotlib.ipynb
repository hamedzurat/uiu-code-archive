**What is Matplotlib?**

Matplotlib is a low level graph plotting library in python that serves as a visualization utility.

Most of the Matplotlib utilities lies under the **pyplot** submodule, and are usually imported under the plt alias

**Plotting x and y points**

The plot() function is used to draw points in a diagram. By default, the plot() function draws a line from point to point.



*   Parameter 1 is an array containing the points on **x-axis**
*   Parameter 2 is an array containing the points on **y-axis**



```python
import matplotlib.pyplot as plt
import numpy as np

x = np.array([0,10])
y = np.array([0,20])

plt.plot(x, y)
plt.show()
```

**Plotting without line**

```python
x = np.array([1,2,3,4])
y = np.array([3,8,-5,10])

plt.plot(x,y, 'o')
plt.show()
```

**Multiple Lines**

**Create Labels and a Title for a Plot**

```python
x = np.array([10,30,50,65,80,90])
y = np.array([25,40,80,105,120,150])

plt.plot(x,y)
plt.xlabel("Blood pulse")
plt.ylabel("Calories burned")
plt.title("Sports Watch Data")
plt.show()
```

**Display Multiple Plots**

```python
x1 = np.array([0,1,2,3])
y1 = np.array([3,8,1,10])
plt.subplot(1,2,1)
plt.plot(x1,y1)

x2 = np.array([0,1,2,3])
y2 = np.array([3,6,3,9])
plt.subplot(1,2,2)
plt.plot(x2,y2)

plt.show()
```

```python
x1 = np.array([0,1,2,3])
y1 = np.array([3,8,1,10])
plt.subplot(2,1,1)
plt.plot(x1,y1)
plt.title("aaa")

x2 = np.array([0,1,2,3])
y2 = np.array([3,6,3,9])
plt.subplot(2,1,2)
plt.plot(x2,y2)
plt.title("bbb")

plt.suptitle("EEE")
plt.show()
```

Bar Plot

```python
x = np.array([1,5,7,10])
y = np.array([10,20,7,15])
plt.bar(x,y)
plt.show()
```

**Pie Chart**

```python
f = ["apple", "cherry", "straw", "pine"]
y = np.array([35,25,15,25])
plt.pie(y, labels=f)
plt.show()
```
